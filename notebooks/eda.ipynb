```python
plt.figure(figsize=(8, 5))

sns.heatmap(
    df[["Sales", "Quantity", "Discount", "Profit"]].corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Between Sales, Quantity, Discount and Profit")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df[["Sales", "Quantity", "Discount", "Profit"]])

plt.title("Outlier Analysis")
plt.xticks(rotation=45)
plt.show()
```
