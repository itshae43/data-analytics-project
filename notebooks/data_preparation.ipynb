```python
# Check for invalid date ranges
print("Invalid Order Dates:", df["Order Date"].isna().sum())
print("Invalid Ship Dates:", df["Ship Date"].isna().sum())
print("Ship date before order date:", (df["Ship Date"] < df["Order Date"]).sum())
```
